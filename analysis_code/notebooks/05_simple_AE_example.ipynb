```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def positional_encoding(ages, d):
    ages = np.array(ages)
    i = np.arange(d // 2)
    angles = ages[:, np.newaxis] / (10000 ** (2 * i / d))
    
    pos_encodings = np.zeros((len(ages), d))
    pos_encodings[:, 0::2] = np.sin(angles)
    pos_encodings[:, 1::2] = np.cos(angles)
    
    return pos_encodings
```

```python
ages = [0, 1, 2, 3, 4, 5, 10, 20, 50]
d = 16  
pe = positional_encoding(ages, d)
print(pe.shape)
print(pe[:3])  
```

```python
plt.figure(figsize=(10, 5))
plt.imshow(pe, aspect='auto', cmap='viridis')
plt.colorbar(label='Value')
plt.xlabel('Encoding dimension')
plt.ylabel('Time (age)')
plt.title('Positional Encoding pattern')
plt.show()
```
